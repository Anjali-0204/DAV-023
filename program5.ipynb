```python


import pandas as pd
import numpy as np

# --------------------------------
# STEP 1: Create Manual Dataset
# --------------------------------

data = {
    "student": ["S1", "S2", "S3", "S4", "S5",
                "S6", "S7", "S8", "S9", "S10"],
    "studyHours": [2, 3, 4, 5, 3, 4, 6, 5, 4, 25]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)


# --------------------------------
# STEP 2: IQR METHOD
# --------------------------------

Q1 = df["studyHours"].quantile(0.25)
Q3 = df["studyHours"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers_iqr = df[
    (df["studyHours"] < lower) |
    (df["studyHours"] > upper)
]

print("\n----- IQR METHOD -----")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)

print("\nIQR Outliers:")
print(outliers_iqr)


# --------------------------------
# STEP 3: Z-SCORE METHOD
# --------------------------------

mean = df["studyHours"].mean()
std = df["studyHours"].std()

df["z_score"] = (df["studyHours"] - mean) / std

# Identify outliers where absolute Z-score > 3
outliers_z = df[abs(df["z_score"]) > 3]

print("\n----- Z-SCORE METHOD -----")
print("Mean:", mean)
print("Standard Deviation:", std)

print("\nZ-Scores:")
print(df[["student", "studyHours", "z_score"]])

print("\nZ-Score Outliers:")
print(outliers_z)


# --------------------------------
# STEP 4: CAPPING METHOD
# --------------------------------

df["cappedHours"] = df["studyHours"].clip(
    lower=lower,
    upper=upper
)

print("\n----- CAPPING METHOD -----")
print("Original and Capped Values:")
print(df[["student", "studyHours", "cappedHours"]])


# --------------------------------
# STEP 5: FINAL COMPARISON
# --------------------------------

print("\n----- FINAL RESULT -----")

print("Original Data:")
print(df["studyHours"].tolist())

print("Capped Data:")
print(df["cappedHours"].tolist())

print("\nTotal IQR Outliers:", len(outliers_iqr))
print("Total Z-Score Outliers:", len(outliers_z))

```
