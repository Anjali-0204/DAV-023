```python
import pandas as pd
import numpy as np

# Step 1: Create a dataset
data = {
    'Name': ['Anu', 'Ravi', 'Sneha', 'Kiran', 'Priya'],
    'Age': [20, 21, np.nan, 22, 23],
    'Salary': [25000, 30000, 35000, np.nan, 40000],
    'City': ['Mysore', 'Bangalore', np.nan, 'Mysore', 'Bangalore'],
    'Gender': ['Female', 'Male', 'Female', 'Male', 'Female']
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Step 2: Handle missing values using Mean
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Step 3: Handle missing values using Median
df['Salary'] = df['Salary'].fillna(df['Salary'].median())

# Step 4: Handle missing values using Forward Fill
df['City'] = df['City'].ffill()

print("\nDataset After Handling Missing Values:")
print(df)

# Step 5: Encode categorical variables
df_encoded = pd.get_dummies(
    df,
    columns=['City', 'Gender'],
    dtype=int
)

print("\nDataset After Encoding Categorical Variables:")
print(df_encoded)

# Step 6: Check missing values
print("\nMissing Values:")
print(df_encoded.isnull().sum())
```
